# TASK - 3

You are working with a dataset of employee salaries, stored in a pandas DataFrame
named data. The dataset includes the following columns: Department, EmployeeID,
Salary, YearsOfExperience, and Age. You need to perform the following tasks using the
groupby() function:






```python
s_data=pd.read_csv("employee_salaries.csv")
s_data
```

1. Write the code to group the data by Department and calculate the mean salary
for each group. What is the result of this operation? What insights can you draw
from the mean salaries of each department?

```python
dept_groupby=s_data.groupby('Department')
dept_groupby['Salary'].mean()
```

**Insights**

- Marketing has the highest average salary (₹85,255), indicating it might be a high-performance team with more seniors.

- Finance shows the lowest average (₹70,509), indicating it having more entry-level employees.

- Engineering lies in the middle (₹76,457), it may have a balanced mix of junior and senior roles.

2. How can you group by multiple columns and apply multiple aggregation
functions? Write the code for this operation and explain the output.


```python
multi_group = s_data.groupby(['Department', 'YearsOfExperience']).agg({
    'Salary': ['mean', 'max', 'min'],
    'Age': ['mean', 'max']
})
multi_group
```

**Insights**

- Salary - mean: Average salary for employees in that department with those years of experience.

- Salary - max: Maximum salary in that group.

- Salary - min: Minimum salary in that group.

- Age - mean: Average age of employees in that group.

- Age - max: Oldest age in that group.

3. How would you filter the grouped data based on a condition (e.g., mean salary
greater than $60,000)? Write the code to perform this filtering and describe the
results.


```python
multi_group[multi_group[('Salary', 'mean')] > 60000]
```

4. How can you define and apply a custom aggregation function to each group?
Write the code for this operation and explain what the results represent.


```python
def salary_range(s):
    return s.max() - s.min()
s_data.groupby('Department')['Salary'].agg(salary_range)
```
